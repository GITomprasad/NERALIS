```python
def get_season(month):
    if month in [12, 1, 2]:
        return "WINTER"
    elif month in [3, 4, 5]:
        return "PRE_MONSOON"
    elif month in [6, 7, 8, 9]:
        return "MONSOON"
    else:
        return "POST_MONSOON"


ner_ml["season"] = ner_ml["event_month"].apply(get_season)

ner_ml["is_monsoon"] = (
    ner_ml["event_month"]
    .isin([6, 7, 8, 9])
    .astype(int)
)

print("Season distribution:")
print(ner_ml["season"].value_counts())

print("\nMonsoon flag:")
print(ner_ml["is_monsoon"].value_counts())
```

```python
import re
import unicodedata

def normalize_text(text):
    if pd.isna(text):
        return None
    
    text = str(text)
    text = unicodedata.normalize("NFKD", text)
    text = "".join(
        char for char in text
        if not unicodedata.combining(char)
    )
    text = text.upper()
    text = re.sub(r"[^A-Z0-9 ]", " ", text)
    text = re.sub(r"\s+", " ", text).strip()
    
    return text


# Normalize district names
district_rainfall_ner["district_normalized"] = (
    district_rainfall_ner["DISTRICT"]
    .apply(normalize_text)
)

# Re-normalize NASA closest locations for consistency
ner_ml["closest_normalized"] = (
    ner_ml["gazeteer_closest_point"]
    .apply(normalize_text)
)

# Build state + location keys
ner_ml["district_match_key"] = (
    ner_ml["state"] + " | " + ner_ml["closest_normalized"]
)

district_rainfall_ner["district_match_key"] = (
    district_rainfall_ner["STATE_UT_NAME"] + " | " +
    district_rainfall_ner["district_normalized"]
)

# Check direct match coverage
district_keys = set(
    district_rainfall_ner["district_match_key"]
)

ner_ml["district_direct_match"] = (
    ner_ml["district_match_key"]
    .isin(district_keys)
)

print("Direct district match coverage:")
print(ner_ml["district_direct_match"].value_counts())

print("\nMatched events:", ner_ml["district_direct_match"].sum())
print(
    "Unmatched events:",
    (~ner_ml["district_direct_match"]).sum()
)
```

```python
from sklearn.model_selection import StratifiedKFold, cross_validate
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.ensemble import RandomForestClassifier

# ============================================================
# BASE FEATURES
# ============================================================

features_without_district = [
    "latitude",
    "longitude",
    "admin_division_population",
    "gazeteer_distance",
    "event_year",
    "event_month",
    "is_monsoon",
    "event_month_rainfall",
    "seasonal_rainfall",
    "ANNUAL",
    "rainfall1_match"
]


# ============================================================
# FIND DISTRICT COLUMNS
# ============================================================

district_columns = [
    col for col in ner_ml.columns
    if col.startswith("district_")
]

print("All district columns:")
print(district_columns)


# ============================================================
# KEEP ONLY NUMERIC DISTRICT FEATURES
# ============================================================

district_features = [
    col for col in district_columns
    if ner_ml[col].dtype.kind in "biufc"
]

print("\nNumeric district features:")
print(district_features)


# ============================================================
# CREATE FEATURE SETS
# ============================================================

features_with_district = (
    features_without_district + district_features
)


# ============================================================
# CHECK FEATURES
# ============================================================

missing_base = [
    col for col in features_without_district
    if col not in ner_ml.columns
]

if missing_base:
    raise KeyError(
        f"Missing base features: {missing_base}"
    )


# ============================================================
# CREATE DATASETS
# ============================================================

X_base = ner_ml[features_without_district]

X_full = ner_ml[features_with_district]


# ============================================================
# VERIFY DATA TYPES
# ============================================================

print("\n==============================================")
print("BASE FEATURE TYPES")
print("==============================================")

print(X_base.dtypes)

print("\n==============================================")
print("FULL FEATURE TYPES")
print("==============================================")

print(X_full.dtypes)


# ============================================================
# STRATIFIED 5-FOLD CROSS VALIDATION
# ============================================================

cv = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)


# ============================================================
# EVALUATION FUNCTION
# ============================================================

def evaluate_features(X_data):

    pipeline = Pipeline([
        (
            "imputer",
            SimpleImputer(strategy="median")
        ),
        (
            "classifier",
            RandomForestClassifier(
                n_estimators=300,
                class_weight="balanced",
                random_state=42,
                n_jobs=-1
            )
        )
    ])

    results = cross_validate(
        pipeline,
        X_data,
        y,
        cv=cv,
        scoring={
            "balanced_accuracy": "balanced_accuracy",
            "macro_f1": "f1_macro"
        },
        n_jobs=-1
    )

    return (
        results["test_balanced_accuracy"].mean(),
        results["test_macro_f1"].mean()
    )


# ============================================================
# EVALUATE BASE FEATURES
# ============================================================

base_bal_acc, base_f1 = evaluate_features(X_base)


# ============================================================
# EVALUATE DISTRICT FEATURES
# ============================================================

full_bal_acc, full_f1 = evaluate_features(X_full)


# ============================================================
# RESULTS
# ============================================================

print("\n==============================================")
print("FEATURE COMPARISON")
print("==============================================")

print("\nWITHOUT district rainfall:")
print("Number of features:", len(features_without_district))
print("Balanced Accuracy:", round(base_bal_acc, 4))
print("Macro F1:", round(base_f1, 4))

print("\nWITH district rainfall:")
print("Number of features:", len(features_with_district))
print("Balanced Accuracy:", round(full_bal_acc, 4))
print("Macro F1:", round(full_f1, 4))


# ============================================================
# IMPROVEMENT
# ============================================================

bal_acc_change = full_bal_acc - base_bal_acc
f1_change = full_f1 - base_f1

print("\n==============================================")
print("IMPROVEMENT")
print("==============================================")

print(
    "Balanced Accuracy change:",
    round(bal_acc_change, 4)
)

print(
    "Macro F1 change:",
    round(f1_change, 4)
)


# ============================================================
# FINAL DECISION
# ============================================================

print("\n==============================================")
print("DECISION")
print("==============================================")

if bal_acc_change > 0 and f1_change > 0:
    print("KEEP district rainfall features.")
    print("Both Balanced Accuracy and Macro F1 improved.")

elif bal_acc_change < 0 and f1_change < 0:
    print("DO NOT keep district rainfall features.")
    print("Both metrics decreased.")

else:
    print("DISTRICT RAINFALL HAS A MIXED EFFECT.")
    print("Check which metric is more important for the final model.")
```

```python
# ============================================================
# PROPER NESTED THRESHOLD VALIDATION
# ============================================================

import numpy as np
import pandas as pd

from sklearn.base import clone
from sklearn.metrics import (
    balanced_accuracy_score,
    f1_score,
    classification_report,
    confusion_matrix
)

# ============================================================
# SETTINGS
# ============================================================

HIGH_THRESHOLDS = np.arange(0.10, 0.61, 0.05)
LOW_THRESHOLDS = np.arange(0.10, 0.61, 0.05)

# Inner CV used ONLY for selecting thresholds
INNER_SPLITS = 3


# ============================================================
# THRESHOLD PREDICTION FUNCTION
# ============================================================

def predict_with_thresholds(
    probabilities,
    high_threshold,
    low_threshold
):
    """
    Decision rule:

    1. HIGH if P(HIGH) >= high_threshold
    2. LOW  if P(LOW)  >= low_threshold
    3. Otherwise MEDIUM

    HIGH gets priority if both HIGH and LOW thresholds
    are satisfied.
    """

    predictions = np.full(
        probabilities.shape[0],
        "MEDIUM",
        dtype=object
    )

    low_mask = (
        probabilities[:, 1] >= low_threshold
    )

    high_mask = (
        probabilities[:, 0] >= high_threshold
    )

    predictions[low_mask] = "LOW"
    predictions[high_mask] = "HIGH"

    return predictions


# ============================================================
# FIND BEST THRESHOLDS USING ONLY TRAINING DATA
# ============================================================

def find_best_thresholds(
    model,
    X_train,
    y_train
):

    # --------------------------------------------------------
    # Inner CV
    # --------------------------------------------------------

    from sklearn.model_selection import StratifiedKFold
    from sklearn.model_selection import cross_val_predict

    inner_cv = StratifiedKFold(
        n_splits=INNER_SPLITS,
        shuffle=True,
        random_state=42
    )

    # --------------------------------------------------------
    # Generate INNER out-of-fold probabilities
    # --------------------------------------------------------

    inner_probabilities = cross_val_predict(
        clone(model),
        X_train,
        y_train,
        cv=inner_cv,
        method="predict_proba",
        n_jobs=-1
    )

    # --------------------------------------------------------
    # Search thresholds
    # --------------------------------------------------------

    best_macro_f1 = -np.inf
    best_balanced_accuracy = -np.inf

    best_high_threshold = 0.50
    best_low_threshold = 0.50

    for high_threshold in HIGH_THRESHOLDS:

        for low_threshold in LOW_THRESHOLDS:

            predictions = predict_with_thresholds(
                inner_probabilities,
                high_threshold,
                low_threshold
            )

            macro_f1 = f1_score(
                y_train,
                predictions,
                average="macro"
            )

            balanced_acc = balanced_accuracy_score(
                y_train,
                predictions
            )

            # Primary objective = Macro F1
            # Tie-breaker = Balanced Accuracy
            if (
                macro_f1 > best_macro_f1
                or (
                    np.isclose(
                        macro_f1,
                        best_macro_f1
                    )
                    and balanced_acc
                    > best_balanced_accuracy
                )
            ):

                best_macro_f1 = macro_f1
                best_balanced_accuracy = balanced_acc

                best_high_threshold = high_threshold
                best_low_threshold = low_threshold

    return (
        best_high_threshold,
        best_low_threshold
    )


# ============================================================
# OUTER CROSS-VALIDATION
# ============================================================

print("==============================================")
print("PROPER NESTED THRESHOLD VALIDATION")
print("==============================================")

print("\nOuter CV folds:", cv.n_splits)
print("Inner CV folds:", INNER_SPLITS)

print("\nStarting validation...\n")


outer_predictions_baseline = []
outer_predictions_threshold = []

outer_true_labels = []

fold_results = []

fold_number = 1


# ============================================================
# OUTER FOLD LOOP
# ============================================================

for train_idx, test_idx in cv.split(X_final, y):

    print(
        f"Processing outer fold {fold_number}/{cv.n_splits}..."
    )

    # --------------------------------------------------------
    # Split data
    # --------------------------------------------------------

    X_train = X_final.iloc[train_idx]
    X_test = X_final.iloc[test_idx]

    y_train = y.iloc[train_idx]
    y_test = y.iloc[test_idx]

    # --------------------------------------------------------
    # Find thresholds ONLY using outer training data
    # --------------------------------------------------------

    (
        best_high_threshold,
        best_low_threshold
    ) = find_best_thresholds(
        best_rf_model,
        X_train,
        y_train
    )

    # --------------------------------------------------------
    # Train model on ENTIRE outer training fold
    # --------------------------------------------------------

    fold_model = clone(best_rf_model)

    fold_model.fit(
        X_train,
        y_train
    )

    # --------------------------------------------------------
    # Predict probabilities on completely unseen fold
    # --------------------------------------------------------

    test_probabilities = (
        fold_model.predict_proba(X_test)
    )

    # --------------------------------------------------------
    # Normal RF prediction
    # --------------------------------------------------------

    baseline_predictions = (
        fold_model.predict(X_test)
    )

    # --------------------------------------------------------
    # Threshold prediction
    # --------------------------------------------------------

    threshold_predictions = (
        predict_with_thresholds(
            test_probabilities,
            best_high_threshold,
            best_low_threshold
        )
    )

    # --------------------------------------------------------
    # Fold metrics
    # --------------------------------------------------------

    baseline_balanced_acc = (
        balanced_accuracy_score(
            y_test,
            baseline_predictions
        )
    )

    baseline_macro_f1 = (
        f1_score(
            y_test,
            baseline_predictions,
            average="macro"
        )
    )

    threshold_balanced_acc = (
        balanced_accuracy_score(
            y_test,
            threshold_predictions
        )
    )

    threshold_macro_f1 = (
        f1_score(
            y_test,
            threshold_predictions,
            average="macro"
        )
    )

    fold_results.append({
        "Fold": fold_number,

        "HIGH Threshold":
            best_high_threshold,

        "LOW Threshold":
            best_low_threshold,

        "Baseline Balanced Accuracy":
            baseline_balanced_acc,

        "Baseline Macro F1":
            baseline_macro_f1,

        "Threshold Balanced Accuracy":
            threshold_balanced_acc,

        "Threshold Macro F1":
            threshold_macro_f1
    })

    # --------------------------------------------------------
    # Store predictions
    # --------------------------------------------------------

    outer_true_labels.extend(
        y_test.tolist()
    )

    outer_predictions_baseline.extend(
        baseline_predictions.tolist()
    )

    outer_predictions_threshold.extend(
        threshold_predictions.tolist()
    )

    fold_number += 1


# ============================================================
# FOLD RESULTS
# ============================================================

fold_results_df = pd.DataFrame(
    fold_results
)

print("\n==============================================")
print("FOLD-BY-FOLD RESULTS")
print("==============================================")

print(
    fold_results_df.to_string(
        index=False,
        float_format=lambda x: f"{x:.4f}"
    )
)


# ============================================================
# MEAN CV PERFORMANCE
# ============================================================

mean_baseline_balanced_accuracy = (
    fold_results_df[
        "Baseline Balanced Accuracy"
    ].mean()
)

mean_baseline_macro_f1 = (
    fold_results_df[
        "Baseline Macro F1"
    ].mean()
)

mean_threshold_balanced_accuracy = (
    fold_results_df[
        "Threshold Balanced Accuracy"
    ].mean()
)

mean_threshold_macro_f1 = (
    fold_results_df[
        "Threshold Macro F1"
    ].mean()
)


# ============================================================
# OVERALL OUTER-CV PERFORMANCE
# ============================================================

overall_baseline_balanced_accuracy = (
    balanced_accuracy_score(
        outer_true_labels,
        outer_predictions_baseline
    )
)

overall_baseline_macro_f1 = (
    f1_score(
        outer_true_labels,
        outer_predictions_baseline,
        average="macro"
    )
)

overall_threshold_balanced_accuracy = (
    balanced_accuracy_score(
        outer_true_labels,
        outer_predictions_threshold
    )
)

overall_threshold_macro_f1 = (
    f1_score(
        outer_true_labels,
        outer_predictions_threshold,
        average="macro"
    )
)


# ============================================================
# FINAL RESULTS
# ============================================================

print("\n==============================================")
print("PROPER NESTED THRESHOLD VALIDATION RESULTS")
print("==============================================")

print("\nMEAN OUTER-CV RESULTS:")

print(
    "Baseline RF Balanced Accuracy:",
    round(
        mean_baseline_balanced_accuracy,
        4
    )
)

print(
    "Baseline RF Macro F1:",
    round(
        mean_baseline_macro_f1,
        4
    )
)

print(
    "Threshold RF Balanced Accuracy:",
    round(
        mean_threshold_balanced_accuracy,
        4
    )
)

print(
    "Threshold RF Macro F1:",
    round(
        mean_threshold_macro_f1,
        4
    )
)


# ============================================================
# OVERALL OUTER-CV RESULTS
# ============================================================

print("\n==============================================")
print("OVERALL OUTER-CV RESULTS")
print("==============================================")

print("\nBaseline RF:")

print(
    "Balanced Accuracy:",
    round(
        overall_baseline_balanced_accuracy,
        4
    )
)

print(
    "Macro F1:",
    round(
        overall_baseline_macro_f1,
        4
    )
)

print("\nThreshold-Tuned RF:")

print(
    "Balanced Accuracy:",
    round(
        overall_threshold_balanced_accuracy,
        4
    )
)

print(
    "Macro F1:",
    round(
        overall_threshold_macro_f1,
        4
    )
)


# ============================================================
# IMPROVEMENT
# ============================================================

balanced_accuracy_change = (
    mean_threshold_balanced_accuracy
    - mean_baseline_balanced_accuracy
)

macro_f1_change = (
    mean_threshold_macro_f1
    - mean_baseline_macro_f1
)

print("\n==============================================")
print("THRESHOLD IMPROVEMENT")
print("==============================================")

print(
    "Balanced Accuracy Change:",
    f"{balanced_accuracy_change:+.4f}"
)

print(
    "Macro F1 Change:",
    f"{macro_f1_change:+.4f}"
)


# ============================================================
# FINAL CLASSIFICATION REPORT
# ============================================================

print("\n==============================================")
print("BASELINE RF CLASSIFICATION REPORT")
print("==============================================")

print(
    classification_report(
        outer_true_labels,
        outer_predictions_baseline,
        digits=4
    )
)


print("\n==============================================")
print("THRESHOLD RF CLASSIFICATION REPORT")
print("==============================================")

print(
    classification_report(
        outer_true_labels,
        outer_predictions_threshold,
        digits=4
    )
)


# ============================================================
# CONFUSION MATRIX — BASELINE
# ============================================================

labels = [
    "HIGH",
    "LOW",
    "MEDIUM"
]

baseline_cm = confusion_matrix(
    outer_true_labels,
    outer_predictions_baseline,
    labels=labels
)

baseline_cm_df = pd.DataFrame(
    baseline_cm,
    index=[
        "Actual HIGH",
        "Actual LOW",
        "Actual MEDIUM"
    ],
    columns=[
        "Predicted HIGH",
        "Predicted LOW",
        "Predicted MEDIUM"
    ]
)

print("\n==============================================")
print("BASELINE RF CONFUSION MATRIX")
print("==============================================")

print(baseline_cm_df)


# ============================================================
# CONFUSION MATRIX — THRESHOLD
# ============================================================

threshold_cm = confusion_matrix(
    outer_true_labels,
    outer_predictions_threshold,
    labels=labels
)

threshold_cm_df = pd.DataFrame(
    threshold_cm,
    index=[
        "Actual HIGH",
        "Actual LOW",
        "Actual MEDIUM"
    ],
    columns=[
        "Predicted HIGH",
        "Predicted LOW",
        "Predicted MEDIUM"
    ]
)

print("\n==============================================")
print("THRESHOLD RF CONFUSION MATRIX")
print("==============================================")

print(threshold_cm_df)


# ============================================================
# THRESHOLD STABILITY
# ============================================================

print("\n==============================================")
print("THRESHOLD STABILITY ACROSS OUTER FOLDS")
print("==============================================")

print(
    "Average HIGH threshold:",
    round(
        fold_results_df[
            "HIGH Threshold"
        ].mean(),
        4
    )
)

print(
    "Average LOW threshold:",
    round(
        fold_results_df[
            "LOW Threshold"
        ].mean(),
        4
    )
)

print(
    "HIGH threshold values:",
    list(
        fold_results_df[
            "HIGH Threshold"
        ]
    )
)

print(
    "LOW threshold values:",
    list(
        fold_results_df[
            "LOW Threshold"
        ]
    )
)


# ============================================================
# FINAL DECISION
# ============================================================

print("\n==============================================")
print("DECISION")
print("==============================================")

if (
    mean_threshold_macro_f1
    > mean_baseline_macro_f1
    and
    mean_threshold_balanced_accuracy
    >= mean_baseline_balanced_accuracy
):

    print(
        "KEEP threshold tuning."
    )

    print(
        "Threshold tuning improved both "
        "Macro F1 and Balanced Accuracy."
    )

elif (
    mean_threshold_macro_f1
    > mean_baseline_macro_f1
):

    print(
        "POSSIBLY KEEP threshold tuning."
    )

    print(
        "Macro F1 improved, but Balanced "
        "Accuracy did not improve."
    )

else:

    print(
        "REJECT threshold tuning."
    )

    print(
        "The improvement did not survive "
        "proper nested validation."
    )
```
